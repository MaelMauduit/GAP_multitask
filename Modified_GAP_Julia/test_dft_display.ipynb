```python
%matplotlib inline
from ase.io import read
import matplotlib.pyplot as plt

def I_want_a_plot(Ecut, color = "b"):
    traj = read(f"bcc_dft_{Ecut}.xyz", index=":")

    print("Nombre de frames :", len(traj))

    volume = []
    energy = []
    for i, atoms in enumerate(traj):
        volume.append(atoms.info["volume"])
        energy.append(atoms.info["dft_energy"])
    plt.plot(volume, energy, c=color)
```

```python
I_want_a_plot(10, "b")
I_want_a_plot(18, "g")
I_want_a_plot(26, "r")
```
